```python
import pandas as pd

all_sheets = pd.read_excel('/content/drive/MyDrive/Cadetx /ev_charging_dataset.xlsx', sheet_name=None)
sessions_df = all_sheets['sessions']

print(sessions_df.shape)
```

```python
sessions_df['month'] = sessions_df['start_timestamp'].dt.month

monthly_demand = sessions_df.groupby('month').size().reset_index(name='session_count')
print(monthly_demand)
```

```python
region_pattern = sessions_df.groupby('station_region').size().reset_index(name='session_count')
print(region_pattern.sort_values('session_count', ascending=False))

charger_type_pattern = sessions_df.groupby('charger_type').size().reset_index(name='session_count')
print(charger_type_pattern.sort_values('session_count', ascending=False))
```

```python
chargers_df = all_sheets['chargers']

charger_type_counts = chargers_df.groupby('charger_type').size().reset_index(name='charger_count')
print(charger_type_counts)
```

```python
charger_type_analysis = charger_type_pattern.merge(charger_type_counts, on='charger_type')
charger_type_analysis['sessions_per_charger'] = charger_type_analysis['session_count'] / charger_type_analysis['charger_count']

print(charger_type_analysis)
```

```python
monthly_demand.to_csv('/content/drive/MyDrive/Cadetx /monthly_demand.csv', index=False)
region_pattern.to_csv('/content/drive/MyDrive/Cadetx /region_pattern.csv', index=False)
charger_type_analysis.to_csv('/content/drive/MyDrive/Cadetx /charger_type_analysis.csv', index=False)

print("Saved!")
```
