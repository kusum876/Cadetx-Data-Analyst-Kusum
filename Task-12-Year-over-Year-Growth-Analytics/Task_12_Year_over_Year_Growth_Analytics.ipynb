```python
import pandas as pd

all_sheets = pd.read_excel('/content/drive/MyDrive/Cadetx /ev_charging_dataset.xlsx', sheet_name=None)
sessions_df = all_sheets['sessions']
stations_df = all_sheets['stations']

print(sessions_df.shape)
```

```python
yearly_kpis = sessions_df.groupby('year').agg(
    total_sessions=('session_id', 'count'),
    total_revenue=('total_cost', 'sum'),
    unique_customers=('customer_id', 'nunique')
).reset_index()

print(yearly_kpis)
```

```python
yearly_kpis['sessions_per_customer'] = yearly_kpis['total_sessions'] / yearly_kpis['unique_customers']
print(yearly_kpis)
```

```python
yearly_kpis.to_csv('/content/drive/MyDrive/Cadetx /yearly_kpis.csv', index=False)
print("Saved!")
```
